```python
import torch
import numpy as np
```

# initializating a tensor

```python
# directly from data
data = [[1, 2], [3, 4]]
x_data = torch.tensor(data)

# from numpy
np_array = np.array(data)
x_np = torch.from_numpy(np_array)

print(f"x_data: \n{x_data}")
print(f"x_np: \n{x_np}")
```

```python
# from another tensor
x_ones = torch.ones_like(x_data)
x_rand = torch.rand_like(x_data, dtype=torch.float)

print(f"x_ones: \n{x_ones}")
print(f"x_rand: \n{x_rand}")
```

```python
# with random or cnstant values
shape = (2,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"rand_tensor: \n{rand_tensor}" )
print(f"ones_tensor: \n{ones_tensor}")
print(f"zeros_tensor: \n{zeros_tensor}")
```

# attributes of a tensor

```python
tensor = torch.rand(3,4)
print(tensor.shape)
print(tensor.dtype)
print(tensor.device)
```

# operations on tensors

```python
# データをCPU -> GPUへ転送
## これが一番モダンな書き方
tensor = torch.rand(3,4)  # cpu
print(tensor.device)
if torch.accelerator.is_available():
    tensor = tensor.to(torch.accelerator.current_accelerator())
print(tensor.device)  # mps:0

## いつも見る書き方
tensor = torch.rand(3,4)  # cpu
print(tensor.device)
device = "cuda" if torch.cuda.is_available() else "cpu"
tensor = tensor.to(device)
print(tensor.device)  # cpu(MACにはgpuはないため)

device = "mps" if torch.mps.is_available() else "cpu"
tensor = tensor.to(device)
print(tensor.device)  # mps:0(MACにはmpsはあるため)
```

```python
tensor = torch.ones(4,4)
print(tensor)
print(f"first raw: {tensor[0]}")
print(f"first column: {tensor[:, 0]}")
print(f"last column: {tensor[..., -1]}")  # tensor[..., -1] is the same as tensor[:, :, -1]

tensor[:,1] = 0
print(tensor)
```

```python
torch.cat([tensor, tensor], dim=1)
# torch.cat([tensor, tensor])
```

```python
# 行列積(Matri Multipication)
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)
y3 = torch.rand_like(y1)  # 事前に作った箱（y3）に入れる。メモリ節約
torch.matmul(tensor, tensor.T, out=y3)

print(y1)
print(y2)
print(y3)
```

```python
# 要素積(Elementwise Product)
z1 = tensor * tensor
z2 = tensor.mul(tensor)
z3 = torch.rand_like(tensor)
torch.mul(tensor, tensor, out=z3)

print(z1)
print(z2)
print(z3)
```

```python
agg = tensor.sum()
print(agg)
agg_item = agg.item()  # convert single element tensor to a python float
print(agg_item)
```

```python
# in-place operations(直接書き換え。メモリを新たに食わない)
# けどbackpropargation中にコレをやるとエラーの原因になるのでやらないのが無難
print(tensor)
tensor.add_(5)
print(tensor)
```

# bridge with numpy

```python
# tensor to numpy array
t = torch.ones(5)
print(f"tensor: {t}")
n = t.numpy()
print(f"numpy array: {n}")
```

```python
# a change in the tensor refletcts in the numpy array
t.add_(1)
print(t)
print(n)
```

```python
# numpy array to tensor
n = np.ones(5)
t = torch.from_numpy(n)
print(t)
```

# additional problems from gemini

```python
# broadcasting
a = torch.rand(3,1)
b = torch.rand(1,4)
c = a + b
print(c.shape)  # (3,1)と(1,4)は互いに拡張されるので、結果は(3,4)となる
```

```python
# flatten
images = torch.rand(8, 1, 28, 28)
print(images.shape)
flattened_images = images.view(8, -1)
print(flattened_images.shape)

flattened_images2 = torch.flatten(images, start_dim=1)
print(flattened_images2.shape)
```

```python
# mask
tensor = torch.randn(5,5)
print(tensor)
tensor[tensor < 0] = 0
print(tensor)
```

```python
# permute
img_torch = torch.rand(3, 100, 100)  # C, H, W
img_torch = img_torch.permute(1, 2, 0)  # H, W, C
print(img_torch.shape)
```
